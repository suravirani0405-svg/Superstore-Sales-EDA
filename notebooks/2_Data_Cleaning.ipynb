### Incorrect Data Types

After inspecting the dataset using `df.info()`, the following observations were made:

| Column | Current Data Type | Correct Data Type | Reason |
|---------|-------------------|-------------------|--------|
| Order Date | str | datetime64[ns] | Required for time-based analysis |

These columns will be converted to the `datetime64[ns]` datatype before performing Exploratory Data Analysis.

- This conversion enables date-based operations such as extracting years, months, and calculating shipping duration.

The `Order Date` and `Ship Date` columns were initially stored as strings (`str`). Strings cannot be used for date-based operations such as extracting years, months, or calculating the number of days between two dates.

# Task 7: Check Outliers

---

## Objective

The objective of this step is to identify outliers in the numerical columns. Outliers are unusual observations that are significantly different from the rest of the data. They are identified but not removed at this stage because they may represent genuine business transactions.

### Does this column contain outliers?

```python
df.select_dtypes(include='number').columns
```

```python
import matplotlib.pyplot as plt

numeric_cols = ['Sales', 'Postal Code']

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.boxplot(df[col].dropna())
    plt.title(f'Boxplot of {col}')
    plt.ylabel(col)
    plt.show()
```

### Sales

**Observation:**

- The Sales column contains several outliers.
- These outliers likely represent high-value customer orders.
- Since high sales are valid business events, these outliers will not be removed during the data cleaning phase.

### Postal Code

**Observation:**

- The Postal Code column may appear to have outliers because postal codes are numerical identifiers.
- However, postal codes are not continuous numerical measurements.
- Therefore, outlier detection is not meaningful for this column, and no action is required.

### Outlier Summary

- Outliers were identified using boxplots.
- Outliers were observed in the Sales column.
- These outliers were retained because they may represent legitimate high-value transactions.
- No outlier analysis was performed for identifier columns such as Row ID and Postal Code.
