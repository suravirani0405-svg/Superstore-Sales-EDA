# Section 2: Univariate Analysis

## A. Sales Distribution

```python
import matplotlib.pyplot as plt
```

##### How are sales distributed?

```python
plt.hist(df['Sales'], bins=30)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.savefig("../images/sales_by_category.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

##### Is the distribution skewed?

```python
df['Sales'].describe()
```

### Observation

- The Sales distribution is highly right-skewed.
- Most transactions involve relatively low sales values.
- A small number of high-value transactions significantly increase the average sales.
- The maximum sales value (22,638.48) is much higher than the median (54.49), indicating the presence of extreme values.

### Business Insight

The business receives a large number of low-value orders, while a few high-value transactions contribute significantly to total revenue. These high-value orders should be analyzed further to understand the products, customers, or regions responsible for them.

### Observation

- **Standard class** dominates and **Same day** mode is rarely used

# Section 3: Bivariate Analysis

## A. Sales by Category

#### Which product category generates the highest sales?

```python
df.groupby('Category')['Sales'].sum().sort_values(ascending=False).plot(kind='bar',color='blue')

plt.title("highest Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.savefig("../images/highest_sales_by_category.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

##### Observation

- Technology generated the highest total sales.
- Furniture ranked second.
- Office Supplies generated the lowest sales.
- Sales are not evenly distributed across product categories.

##### Business Insight

Technology products contribute the largest share of revenue. The company should continue investing in this category while exploring strategies to improve sales in lower-performing categories.

## B. Sales by Region

#### Which region generates the highest sales?

```python
df.groupby('Region')['Sales'].sum().sort_values(ascending=False).plot(kind='bar')

plt.title("highest Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.savefig("../images/highest_sales_by_region.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation

- West generated the highest sales.
- East ranked second.
- South contributed the lowest sales.
- Sales vary considerably across regions.

#### Business Insight

High-performing regions should continue receiving strategic investment, while low-performing regions require targeted marketing and customer acquisition efforts.

## C. Sales by Customer Segment

#### Which customer segment contributes the highest revenue?

```python
df.groupby('Segment')['Sales'].sum().sort_values(ascending=False).plot(kind='bar')

plt.title("highest revenue by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.savefig("../images/highest_revenue_by_customer_segment.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

##### Observation

- Consumer customers generated the highest sales.
- Corporate customers ranked second.
- Home Office customers contributed the least revenue.

##### Business Insight

The Consumer segment is the primary revenue source. Marketing campaigns should continue targeting this segment while developing strategies to increase sales among Corporate and Home Office customers.

## D. Sales by Ship Mode

#### Does shipping mode affect sales?

```python
df.groupby('Ship Mode')['Sales'].sum().sort_values(ascending=False).plot(kind='bar')

plt.title("Total Sales by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Total Sales")

plt.xticks(rotation=90)
plt.savefig("../images/total_sales_by_ship.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

##### Observation

- Standard Class generated the highest sales.
- Second Class ranked second.
- Same Day shipping contributed the lowest sales.

##### Business Insight

Most customers prefer Standard Class shipping. The company should continue optimizing this shipping option while evaluating the demand for premium shipping services.

## E. Top 10 States by Sales

#### Which states contribute the highest sales?

##### Observation

- California generated the highest sales.
- New York and Texas also contributed significantly.
- The top 10 states account for a major share of total revenue.

##### Business Insight

The company has a strong presence in a few key states. Expanding successful strategies from these states to other regions may increase overall sales.

## F. Top 10 Cities by Sales

#### Which cities generate the highest sales?

```python
df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(10).plot(kind='bar')

plt.title("Top 10 Cities by Sales")
plt.xlabel("City")
plt.ylabel("Total Sales")

plt.xticks(rotation=90)
plt.savefig("../images/top_10_cities_by_sales.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

# Section 4: Time-Series Analysis

## A. Sales by Year

##### Which year had highest sales?

```python
df['Order Date'].dtype
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format='%d/%m/%Y'
)
```

```python

df.groupby(df['Order Date'].dt.year)['Sales'].sum().plot(kind='line',marker='o')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Sales by year') 
plt.grid()
plt.savefig("../images/sales_by_year.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation

- Sales showed a slight decline from **2015** to **2016**.
- From **2016 onwards**, sales increased significantly.
- **2018** recorded the highest total sales among all four years.
- The overall trend indicates strong business growth during the later years of the dataset.

#### Business Insight

The continuous increase in sales from 2016 to 2018 suggests that the business expanded successfully during this period. Understanding the factors that contributed to this growth can help the company maintain and improve future performance.

## B. Sales by Month

##### Which months perform best?

```python
df.groupby(df['Order Date'].dt.month)['Sales'].sum().plot(kind='line',marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1,13),labels=['jan','feb','march','april','may','june','july','august','sept','oct','nov','dec'])
plt.grid(True)
plt.savefig("../images/sales_by_month.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation

- Sales fluctuate considerably across different months.
- **November** recorded the highest monthly sales.
- **December** had the second-highest sales, while **October** also showed strong performance.
- **February** recorded the lowest sales among all months.
- Sales generally increase during the last quarter of the year (October–December), indicating a seasonal trend.

#### Business Insight

The significant increase in sales during the final quarter suggests strong seasonal demand, likely driven by holiday shopping or year-end promotions. The company should ensure adequate inventory, staffing, and marketing efforts during these months to maximize revenue.

## C. Orders by Month

##### Count number of orders.

```python
df.groupby(df['Order Date'].dt.month)['Sales'].nunique().plot(kind='line',marker='o')

plt.title("Number of Orders by Month")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(range(1,13),labels=['jan','feb','march','april','may','june','july','august','sept','oct','nov','dec'])
plt.grid(True)
plt.savefig("../images/orders_by_month.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation

- The number of customer orders varies across different months.
- **November** recorded the highest number of orders, followed by **September** and **December**.
- **February** had the lowest number of orders.
- Order volume generally increases during the second half of the year, especially in the final quarter (October–December).
- The trend suggests seasonal fluctuations in customer purchasing behavior.

#### Business Insight

The company experiences peak customer demand during the final quarter of the year. Proper inventory planning, workforce allocation, and marketing campaigns during these months can help maximize sales and maintain service quality.

## D. Orders by year

##### Count yearly orders.

```python

df.groupby(df['Order Date'].dt.year)['Sales'].count().plot(kind='line',marker='o')
plt.title("Number of Orders by Year")
plt.xlabel("Year")
plt.ylabel("Number of Orders")
plt.savefig("../images/orders_by_year.png",
            dpi=300,
            bbox_inches="tight")
plt.grid()
plt.show()
```

#### Observation

- The number of customer orders increased every year from **2015** to **2018**.
- **2018** recorded the highest number of orders.
- The most significant growth occurred between **2017** and **2018**.
- The consistent increase in yearly orders indicates growing customer demand and business expansion.

#### Business Insight

The steady increase in yearly orders reflects positive business growth and an expanding customer base. The company should continue focusing on customer acquisition and retention strategies to sustain this upward trend.
